# 02 - Data Cleaning

Notebook to document and demonstrate the reproducible cleaning pipeline.

```python
import os
import sys
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..', 'src')))

import pandas as pd
import numpy as np

from data_loader import load_raw_dataset, inspect_dataset, format_inspection_table, save_processed_dataset
from data_cleaning import (
    clean_dataset,
    validate_required_columns,
    remove_duplicates,
    strip_whitespace_text_columns,
    normalize_categoricals,
    parse_dates,
    drop_invalid_rows,
    derive_resolution_metrics,
    save_data_quality_report,
    DataQualityReport,
)
```

```python
raw = load_raw_dataset()
raw_inspect = inspect_dataset(raw)
validate_required_columns(raw)
print(f"Raw rows: {len(raw)}")
```

```python
report = DataQualityReport()
report.original_rows = len(raw)
report.missing_before = raw.isnull().sum().to_dict()

df = raw.copy()
df = remove_duplicates(df, report)
df = strip_whitespace_text_columns(df, report)
df = normalize_categoricals(df, report)
df = parse_dates(df, report)
df = drop_invalid_rows(df, report)
df = derive_resolution_metrics(df, report)

report.cleaned_rows = len(df)
report.missing_after = df.isnull().sum().to_dict()
```

```python
print(report.to_text())
```

```python
cleaned_inspect = inspect_dataset(df)
print('\nCleaned dataset:')
print(format_inspection_table(cleaned_inspect))
```

```python
save_processed_dataset(df)
save_data_quality_report(report, raw_inspect, cleaned_inspect)
print('Cleaned data and report written.')
```
