# 01 - Data Exploration

Notebook for initial inspection and validation of the raw IT support ticket dataset.

```python
import os
import sys
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..', 'src')))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

sns.set_theme(style='whitegrid')
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)

from data_loader import load_raw_dataset, inspect_dataset, format_inspection_table
```

```python
df = load_raw_dataset()
inspection = inspect_dataset(df)
print(format_inspection_table(inspection))
```

```python
df.describe(include='all', datetime_is_numeric=True).T.fillna('')
```

```python
cat_cols = [c for c in df.columns if df[c].dtype == 'object' and df[c].nunique() <= 25]
for c in cat_cols:
    print(f"\n=== {c} ===")
    print(df[c].value_counts(dropna=False).head(15))
```

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print('Numerical columns:', num_cols)
for c in num_cols:
    print(f"\n{c}:")
    print(df[c].describe())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
if 'Ticket Type' in df.columns:
    df['Ticket Type'].value_counts().plot.bar(ax=axes[0], color='steelblue', title='Tickets by Type')
    axes[0].tick_params(axis='x', rotation=45)
if 'Ticket Priority' in df.columns:
    df['Ticket Priority'].value_counts().plot.bar(ax=axes[1], color='darkorange', title='Tickets by Priority')
    axes[1].tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()
```

```python
print('Duplicated rows:', df.duplicated().sum())
if 'Ticket ID' in df.columns:
    print('Duplicated Ticket IDs:', df.duplicated(subset=['Ticket ID']).sum())
```
